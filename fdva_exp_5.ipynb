```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("experiment_5_data.csv")

print(df.head())

# Line Chart
plt.plot(df["Student_ID"], df["Marks"], label="Marks")
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.legend()
plt.show()

# Bar Chart
plt.bar(df["Student_ID"][:10], df["Marks"][:10])
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.title("Marks of First 10 Students")
plt.show()

# Histogram
plt.hist(df["Marks"], bins=10, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Student Marks")
plt.show()
```
